## Things worth remembering

- **`.loc` includes the last value, `.iloc` does not.** `df.loc[10:15]` gives 6 rows, `df.iloc[10:15]` gives 5.
- **Use `&` and `|`, not `and` and `or`,** and keep each condition in its own brackets.
- **Two sets of brackets** whenever you pass a list: `total[[0, 10, 50]]` for positions, `df[["City", "Total"]]` for columns.
- **`np.sort` only sorts ascending.** Add `[::-1]` to reverse it.
- **`argmax`, `argmin` and `argsort` return positions, not values.** Anything starting with `arg` gives a position.
- **`axis=0` goes down the rows** and gives one value per column. **`axis=1` goes across the columns** and gives one value per row.
- **`groupby` alone shows nothing.** It only produces a result once you add `.mean()`, `.sum()` or similar.
- **`jointplot`, `pairplot` and `catplot` create their own figure,** so do not call `plt.figure()` before them.
